## Model Specification

```python
import os
from openai import OpenAI

# Your Serving Endpoint name, available in UI
ENDPOINT_NAME = "databricks-gpt-oss-120b"

# Set environment variables in Python
os.environ["DATABRICKS_HOST"] = "Enter workspace URL"
os.environ["DATABRICKS_TOKEN"] = "Enter Developer token"

host = os.environ["DATABRICKS_HOST"].rstrip("/")
token = os.environ["DATABRICKS_TOKEN"]

# Databricks Model Serving is OpenAI-compatible; point the client at your workspace.
# IMPORTANT: base_url must include /serving-endpoints
client = OpenAI(
    api_key=token,
    base_url=f"{host}/serving-endpoints",
)
```

## Sample Inference

```python
resp = client.chat.completions.create(
    model=ENDPOINT_NAME,  # model == your Databricks Serving Endpoint name
    messages=[
        {"role": "system", "content": "You are a general knowledge assistant."},
        {"role": "user", "content": "Who is Roger Federer?"},
    ],
    temperature=0.2,
    max_tokens=250,
)

print(resp.choices[0].message.content)
```

## Listing Available Models

```python
import os
import requests

resp = requests.get(
    f"{host}/api/2.0/serving-endpoints",
    headers={
        "Authorization": f"Bearer {token}",
        "Content-Type": "application/json"
    }
)

resp.raise_for_status()
data = resp.json()

for endpoint in data.get("endpoints", []):
    print(endpoint["name"])
```
